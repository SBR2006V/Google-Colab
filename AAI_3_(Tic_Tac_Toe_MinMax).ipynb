```python
# ==========================================
# TIC-TAC-TOE USING MINIMAX ALGORITHM
# Human = X
# AI = O
# ==========================================


# ------------------------------------------
# 1. CREATE THE BOARD
# ------------------------------------------

board = [" " for _ in range(9)]


# ------------------------------------------
# 2. DISPLAY THE BOARD
# ------------------------------------------

def display_board():
    print()
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("---------")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("---------")
    print(board[6] + " | " + board[7] + " | " + board[8])
    print()


# ------------------------------------------
# 3. CHECK WINNER
# ------------------------------------------

def check_winner(player):

    winning_positions = [
        [0, 1, 2],  # First row
        [3, 4, 5],  # Second row
        [6, 7, 8],  # Third row

        [0, 3, 6],  # First column
        [1, 4, 7],  # Second column
        [2, 5, 8],  # Third column

        [0, 4, 8],  # Main diagonal
        [2, 4, 6]   # Other diagonal
    ]

    for position in winning_positions:

        if all(board[i] == player for i in position):
            return True

    return False


# ------------------------------------------
# 4. CHECK WHETHER BOARD IS FULL
# ------------------------------------------

def board_full():
    return " " not in board


# ------------------------------------------
# 5. HUMAN MOVE
# ------------------------------------------

def human_move():

    while True:

        position = int(input("Enter your position (1-9): "))

        if position < 1 or position > 9:
            print("Please enter a number between 1 and 9.")
            continue

        index = position - 1

        if board[index] == " ":

            board[index] = "X"
            break

        else:
            print("Position already occupied. Try again.")


# ------------------------------------------
# 6. MINIMAX ALGORITHM
# ------------------------------------------

def minimax(is_maximizing):

    # --------------------------------------
    # Check terminal states
    # --------------------------------------

    # AI wins
    if check_winner("O"):
        return 1

    # Human wins
    if check_winner("X"):
        return -1

    # Draw
    if board_full():
        return 0


    # --------------------------------------
    # MAX: AI's turn
    # --------------------------------------

    if is_maximizing:

        best_score = -float("inf")

        for i in range(9):

            if board[i] == " ":

                # Make AI move
                board[i] = "O"

                # Find score of this move
                score = minimax(False)

                # Undo the move
                board[i] = " "

                # Choose maximum score
                best_score = max(best_score, score)

        return best_score


    # --------------------------------------
    # MIN: Human's turn
    # --------------------------------------

    else:

        best_score = float("inf")

        for i in range(9):

            if board[i] == " ":

                # Make human move
                board[i] = "X"

                # Find score of this move
                score = minimax(True)

                # Undo the move
                board[i] = " "

                # Choose minimum score
                best_score = min(best_score, score)

        return best_score


# ------------------------------------------
# 7. AI MOVE USING MINIMAX
# ------------------------------------------

def ai_move():

    best_score = -float("inf")
    best_move = None

    for i in range(9):

        if board[i] == " ":

            # Try AI move
            board[i] = "O"

            # Ask Minimax for the score
            score = minimax(False)

            # Undo move
            board[i] = " "

            # Check whether this is the best move
            if score > best_score:

                best_score = score
                best_move = i

    # Make the best move
    board[best_move] = "O"


# ------------------------------------------
# 8. MAIN GAME
# ------------------------------------------

while True:

    # Display board
    display_board()

    # Human's turn
    human_move()

    # Check if human wins
    if check_winner("X"):

        display_board()
        print("You win!")
        break

    # Check for draw
    if board_full():

        display_board()
        print("Draw!")
        break

    # AI's turn
    print("AI is thinking...")
    ai_move()

    # Check if AI wins
    if check_winner("O"):

        display_board()
        print("AI wins!")
        break

    # Check for draw
    if board_full():

        display_board()
        print("Draw!")
        break
```
