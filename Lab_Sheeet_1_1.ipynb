```python
text = str(input("Enter a string: "))
vowels = 0
consonants = 0
for ch in text:
 if ch.isalpha():
  if ch in "aeiou":
    vowels += 1
  else:
    consonants += 1
print("Vowels =", vowels)
print("Consonants =", consonants)
```

```python
student = {
 "Rahul": 78,
 "Amit": 92,
 "Sneha": 85,
 "Priya": 88
}
print("Sorted by Keys:")
for key in sorted(student.keys()):
 print(key, ":", student[key])
print("\nSorted by Values:")
for key, value in sorted(student.items(), key=lambda item: item[1]):
 print(key, ":", value)
```
