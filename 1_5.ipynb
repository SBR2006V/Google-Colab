```python
import numpy as np
data = {
 "Roll_No": [101, 102, 103, 104, 105],
 "Name": ["Rahul", "Priya", "Amit", "Sneha", "Riya"],
 "Marks": [95, 85, 72, 65, 45]
}
df = pd.DataFrame(data)
conditions = [
 df["Marks"] >= 90,
 df["Marks"] >= 80,
 df["Marks"] >= 70,
 df["Marks"] >= 60
]
grades = ["A+", "A", "B", "C"]
df["Grade"] = np.select(conditions, grades, default="F")
print("Student Data with Grades:")
print(df)
```

```python
data = {
 "Roll_No": [101, 102, 103, 104, 105, 106, 107],
 "Name": ["Rahul", "Priya", "Amit", "Sneha", "Riya", "Karan", "Neha"],
 "Department": ["CSE", "IT", "CSE", "ECE", "CSE", "IT", "ECE"],
 "Marks": [85, 92, 76, 88, 95, 72, 81]
}
df = pd.DataFrame(data)
result = df.groupby("Department").agg(
 Average_Marks=("Marks", "mean"),
 Maximum_Marks=("Marks", "max"),
Minimum_Marks=("Marks", "min"),
 Total_Students=("Roll_No", "count")
)
print("Department-wise Statistics:")
print(result)
```

```python
data = {
 "Name": ["Rahul", "Priya", "Amit", "Sneha", "Riya"],
 "Marks": [85, np.nan, 76, 90, np.nan],
 "Attendance": [90, 85, np.nan, 88, 92]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
print("\nMissing Values:")
print(df.isnull())
print("\nNumber of Missing Values:")
print(df.isnull().sum())
df_mean = df.copy()
df_mean["Marks"] = df_mean["Marks"].fillna(df_mean["Marks"].mean())
print("\nAfter Filling Marks with Mean:")
print(df_mean)
df_median = df.copy()
df_median["Attendance"] = df_median["Attendance"].fillna(
 df_median["Attendance"].median()
)
print("\nAfter Filling Attendance with Median:")
print(df_median)
df_drop = df.dropna()
print("\nAfter Dropping Rows with Missing Values:")
print(df_drop)
```

```python
students = pd.DataFrame({
 "Roll_No": [101, 102, 103, 104],
 "Name": ["Rahul", "Priya", "Amit", "Sneha"],
 "Department": ["CSE", "IT", "CSE", "ECE"]
})
results = pd.DataFrame({
 "Roll_No": [101, 102, 103, 104],
 "Marks": [85, 92, 76, 88],
 "Grade": ["A", "A+", "B", "A"]
})
print("Student Details:")
print(students)
print("\nExamination Results:")
print(results)
merged_df = pd.merge(students, results, on="Roll_No")
print("\nMerged DataFrame:")
print(merged_df)
```
