```python
import numpy as np
```

```python
numbers = input("Enter number seperated by spaces: ")
arr = np.array([int(x) for x in numbers.split()])
print("Array:", arr)
print("Data Type:", arr.dtype)
print("Shape:", arr.shape)
print("Size:", arr.size)
```

```python
marks = np.array([
 [85, 90, 78],
 [75, 88, 92],
 [90, 85, 89],
 [70, 80, 75],
 [95, 92, 96]
])
print("Marks of 5 Students:")
print(marks)
print("Dimensions:", marks.ndim)
print("Shape:", marks.shape)
```

```python
print("Array using arange():")
print(np.arange(1, 11))
print("\nArray using linspace():")
print(np.linspace(0, 1, 5))
print("\nZero Matrix:")
print(np.zeros((3, 3), dtype=int))
print("\nOnes Matrix:")
print(np.ones((3, 3), dtype=int))
print("\nIdentity Matrix:")
print(np.eye(3, dtype=int))
```

```python
a = np.array([10, 20, 30, 40])
b = np.array([2, 4, 5, 8])
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
A = np.array([
 [1, 2],
 [3, 4]
])
B = np.array([
 [5, 6],
 [7, 8]
])
print("Matrix A:")
print(A)
print("\nMatrix B:")
print(B)
print("\nMatrix Multiplication:")
print(np.matmul(A, B))
```
